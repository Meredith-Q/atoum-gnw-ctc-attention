# RTM–WB GNW Mean-Field Notebook

The mean-field model is now the main GNW integration workflow. Open [`RTM-WB GNW Integration.ipynb`](RTM-WB%20GNW%20Integration.ipynb) and run it from top to bottom.

The main notebook represents each anatomical area with an excitatory rate (RTM population) and an inhibitory rate (WB population), while retaining FLN connectivity, V4 drive, workspace ignition, local recurrence, and workspace feedback. It runs on a 1 ms rate grid and does not simulate individual spikes.

The previous cell-by-cell Brian2 implementation is preserved in [`RTM-WB GNW Spiking Reference.ipynb`](RTM-WB%20GNW%20Spiking%20Reference.ipynb).
